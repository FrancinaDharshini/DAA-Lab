```python
import heapq

# Directed weighted graph
graph = {
    0: [(1, 4), (2, 2)],
    1: [(2, 1), (3, 5)],
    2: [(1, 1), (3, 8), (4, 10)],
    3: [(4, 2), (5, 6)],
    4: [(5, 3)],
    5: []
}


def dijkstra(graph, source):
    # Initial distances
    distance = {v: float('inf') for v in graph}
    parent = {v: None for v in graph}

    distance[source] = 0

    # Priority queue: (distance, vertex)
    pq = [(0, source)]

    while pq:
        current_dist, current = heapq.heappop(pq)

        # Ignore outdated entries
        if current_dist > distance[current]:
            continue

        for neighbour, weight in graph[current]:
            new_dist = current_dist + weight

            if new_dist < distance[neighbour]:
                distance[neighbour] = new_dist
                parent[neighbour] = current
                heapq.heappush(pq, (new_dist, neighbour))

    return distance, parent


def get_path(parent, destination):
    path = []

    while destination is not None:
        path.append(destination)
        destination = parent[destination]

    path.reverse()
    return path


# Source vertex
source = 0

distance, parent = dijkstra(graph, source)

print("=" * 60)
print("       DIJKSTRA SHORTEST PATH ALGORITHM")
print("=" * 60)

print("\nSource Vertex : 0")

print("\nDirected Weighted Edges:")
for u in graph:
    for v, w in graph[u]:
        print(f"  {u} -> {v}  (weight = {w})")

print("\n" + "-" * 60)
print(f"{'Destination':<15}{'Distance':<15}{'Shortest Path'}")
print("-" * 60)

for vertex in range(len(graph)):
    path = get_path(parent, vertex)

    if distance[vertex] == float('inf'):
        print(f"{vertex:<15}{'INF':<15}{'No path'}")
    else:
        path_string = " -> ".join(map(str, path))
        print(f"{vertex:<15}{distance[vertex]:<15}{path_string}")

print("-" * 60)
```
