```python
# Matrix Chain Multiplication using Dynamic Programming

def matrix_chain_order(p):
    n = len(p) - 1

    # dp[i][j] = minimum multiplication cost for Ai...Aj
    dp = [[0] * n for _ in range(n)]

    # split[i][j] stores the position where the chain is split
    split = [[0] * n for _ in range(n)]

    # Chain length
    for length in range(2, n + 1):
        for i in range(n - length + 1):
            j = i + length - 1
            dp[i][j] = float('inf')

            for k in range(i, j):
                cost = (
                    dp[i][k]
                    + dp[k + 1][j]
                    + p[i] * p[k + 1] * p[j + 1]
                )

                if cost < dp[i][j]:
                    dp[i][j] = cost
                    split[i][j] = k

    return dp, split


def print_parenthesization(split, i, j):
    if i == j:
        return f"A{i + 1}"

    k = split[i][j]

    left = print_parenthesization(split, i, k)
    right = print_parenthesization(split, k + 1, j)

    return f"({left} × {right})"


# Example:
# A1 = 10×30
# A2 = 30×5
# A3 = 5×60
# A4 = 60×10
# A5 = 10×20

dimensions = [10, 30, 5, 60, 10, 20]

dp, split = matrix_chain_order(dimensions)

n = len(dimensions) - 1
optimal_order = print_parenthesization(split, 0, n - 1)

print("=" * 55)
print("       MATRIX CHAIN MULTIPLICATION")
print("=" * 55)

print("\nMatrix Dimensions:")
for i in range(n):
    print(f"A{i + 1} = {dimensions[i]} × {dimensions[i + 1]}")

print("\nOptimal Parenthesization:")
print(optimal_order)

print("\nMinimum Scalar Multiplications:")
print(dp[0][n - 1])

print("\nResult:")
print(f"Minimum cost = {dp[0][n - 1]} scalar multiplications")
print(f"Best order   = {optimal_order}")
print("=" * 55)
```
