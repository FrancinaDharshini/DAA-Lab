```python
# ============================================================
# MINIMUM AND MAXIMUM: DIVIDE & CONQUER vs LINEAR
# ============================================================

# ---------- Naive Linear Approach ----------
def linear_min_max(arr):
    minimum = arr[0]
    maximum = arr[0]
    comparisons = 0

    for i in range(1, len(arr)):
        # Compare for minimum
        comparisons += 1
        if arr[i] < minimum:
            minimum = arr[i]

        # Compare for maximum
        comparisons += 1
        if arr[i] > maximum:
            maximum = arr[i]

    return minimum, maximum, comparisons


# ---------- Divide and Conquer Approach ----------
def divide_conquer(arr, left, right):
    # Only one element
    if left == right:
        return arr[left], arr[left], 0

    # Two elements
    if right == left + 1:
        comparisons = 1

        if arr[left] < arr[right]:
            return arr[left], arr[right], comparisons
        else:
            return arr[right], arr[left], comparisons

    # Divide
    mid = (left + right) // 2

    min_left, max_left, comp_left = divide_conquer(
        arr, left, mid
    )

    min_right, max_right, comp_right = divide_conquer(
        arr, mid + 1, right
    )

    comparisons = comp_left + comp_right

    # Compare two minimums
    comparisons += 1
    minimum = min_left if min_left < min_right else min_right

    # Compare two maximums
    comparisons += 1
    maximum = max_left if max_left > max_right else max_right

    return minimum, maximum, comparisons


# ============================================================
# TEST DATA
# ============================================================

arr = [42, 17, 89, 3, 56, 71, 8, 94, 25, 63, 11, 38]


# Linear approach
linear_min, linear_max, linear_comp = linear_min_max(arr)

# Divide and Conquer approach
dc_min, dc_max, dc_comp = divide_conquer(
    arr, 0, len(arr) - 1
)


# ============================================================
# DISPLAY RESULTS
# ============================================================

print("=" * 65)
print("       MINIMUM & MAXIMUM : DAA COMPARISON")
print("=" * 65)

print("\nInput Array:")
print(arr)

print("\n" + "-" * 65)

print("NAIVE LINEAR APPROACH")
print("-" * 65)
print("Minimum Element       :", linear_min)
print("Maximum Element       :", linear_max)
print("Number of Comparisons :", linear_comp)

print("\n" + "-" * 65)

print("DIVIDE AND CONQUER APPROACH")
print("-" * 65)
print("Minimum Element       :", dc_min)
print("Maximum Element       :", dc_max)
print("Number of Comparisons :", dc_comp)

print("\n" + "-" * 65)

print("COMPARISON SUMMARY")
print("-" * 65)
print(f"Linear Comparisons    : {linear_comp}")
print(f"Divide & Conquer      : {dc_comp}")
print(f"Comparisons Saved     : {linear_comp - dc_comp}")

print("\n" + "=" * 65)
```
