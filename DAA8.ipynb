```python
INF = float('inf')

# Cost matrix
cost = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

n = len(cost)


def first_min(city):
    """Find the cheapest outgoing edge from a city."""
    values = [cost[city][j] for j in range(n) if j != city]
    return min(values)


def second_min(city):
    """Find the second-cheapest outgoing edge."""
    values = sorted(cost[city][j] for j in range(n) if j != city)
    return values[1]


def branch_and_bound(path, visited, current_cost, bound):
    global best_cost, best_path

    level = len(path)

    # All cities visited
    if level == n:
        last = path[-1]
        start = path[0]

        if cost[last][start] != 0:
            total = current_cost + cost[last][start]

            if total < best_cost:
                best_cost = total
                best_path = path[:] + [start]

        return

    current = path[-1]

    # Try every unvisited city
    for next_city in range(n):

        if not visited[next_city] and cost[current][next_city] != 0:

            new_cost = current_cost + cost[current][next_city]

            # Calculate a lower bound
            new_bound = bound

            if level == 1:
                new_bound -= (first_min(current) + first_min(next_city)) / 2
            else:
                new_bound -= (second_min(current) + first_min(next_city)) / 2

            # Continue only if promising
            if new_cost + new_bound < best_cost:

                visited[next_city] = True
                path.append(next_city)

                branch_and_bound(
                    path,
                    visited,
                    new_cost,
                    new_bound
                )

                path.pop()
                visited[next_city] = False


def solve_tsp():
    global best_cost, best_path

    best_cost = INF
    best_path = []

    # Initial lower bound
    bound = 0

    for city in range(n):
        bound += first_min(city) + second_min(city)

    bound /= 2

    visited = [False] * n
    visited[0] = True

    branch_and_bound(
        [0],
        visited,
        0,
        bound
    )

    return best_path, best_cost


# Solve the problem
tour, minimum_cost = solve_tsp()

print("===== TSP USING BRANCH AND BOUND =====")
print()

print("Cost Matrix:")
for row in cost:
    print(row)

print("\nOptimal Tour:")
print(" -> ".join("City " + str(city + 1) for city in tour))

print("\nMinimum Tour Cost:", minimum_cost)
```
