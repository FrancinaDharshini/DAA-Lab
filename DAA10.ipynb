```python
import random
import time
import sys

sys.setrecursionlimit(30000)

N = 10000


# =========================================================
# DETERMINISTIC QUICK SORT
# Pivot = Last Element
# =========================================================

def deterministic_partition(arr, low, high, counter):
    pivot = arr[high]
    i = low - 1

    for j in range(low, high):
        counter[0] += 1

        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1


def deterministic_quick_sort(arr, low, high, counter):
    if low < high:
        p = deterministic_partition(arr, low, high, counter)

        deterministic_quick_sort(arr, low, p - 1, counter)
        deterministic_quick_sort(arr, p + 1, high, counter)


# =========================================================
# RANDOMIZED QUICK SORT
# Pivot = Random Element
# =========================================================

def randomized_partition(arr, low, high, counter):
    random_index = random.randint(low, high)

    # Move random pivot to last position
    arr[random_index], arr[high] = arr[high], arr[random_index]

    pivot = arr[high]
    i = low - 1

    for j in range(low, high):
        counter[0] += 1

        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = arr[high], arr[i + 1]

    return i + 1


def randomized_quick_sort(arr, low, high, counter):
    if low < high:
        p = randomized_partition(arr, low, high, counter)

        randomized_quick_sort(arr, low, p - 1, counter)
        randomized_quick_sort(arr, p + 1, high, counter)


# =========================================================
# PERFORMANCE TEST
# =========================================================

def test_algorithms(data):
    # Deterministic Quick Sort
    arr1 = data.copy()
    comparisons1 = [0]

    start = time.perf_counter()
    deterministic_quick_sort(arr1, 0, len(arr1) - 1, comparisons1)
    time1 = time.perf_counter() - start

    # Randomized Quick Sort
    arr2 = data.copy()
    comparisons2 = [0]

    start = time.perf_counter()
    randomized_quick_sort(arr2, 0, len(arr2) - 1, comparisons2)
    time2 = time.perf_counter() - start

    return time1, comparisons1[0], time2, comparisons2[0]


# =========================================================
# FOUR ARRAY CONFIGURATIONS
# =========================================================

random.seed(42)

configurations = {
    "Random Array": random.sample(range(1, N * 10), N),

    "Sorted Array": list(range(N)),

    "Reverse Sorted": list(range(N, 0, -1)),

    "Many Duplicates": [random.randint(1, 50) for _ in range(N)]
}


# =========================================================
# MAIN PROGRAM
# =========================================================

print("=" * 72)
print("     QUICK SORT PERFORMANCE: DETERMINISTIC vs RANDOMIZED")
print("=" * 72)
print(f"Array Size: {N}")
print()

print(f"{'Configuration':<20} {'Det.Time':>12} {'Det.Comp':>12} "
      f"{'Rand.Time':>12} {'Rand.Comp':>12}")
print("-" * 72)

for name, data in configurations.items():

    t1, c1, t2, c2 = test_algorithms(data)

    print(f"{name:<20} "
          f"{t1:>10.5f}s "
          f"{c1:>12} "
          f"{t2:>10.5f}s "
          f"{c2:>12}")

print("-" * 72)
print("Lower time and fewer comparisons indicate better performance.")
```
