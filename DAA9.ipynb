```python
import math

weights = [0.5, 0.7, 0.3, 0.9, 0.2, 0.6, 0.8, 0.4, 0.1, 0.5]
CAPACITY = 1.0


# -------------------------------------------------
# First Fit
# -------------------------------------------------
def first_fit(items):
    bins = []

    for item in items:
        placed = False

        for i in range(len(bins)):
            if bins[i] + item <= CAPACITY:
                bins[i] += item
                placed = True
                break

        if not placed:
            bins.append(item)

    return bins


# -------------------------------------------------
# Best Fit
# -------------------------------------------------
def best_fit(items):
    bins = []

    for item in items:
        best_index = -1
        smallest_space = CAPACITY + 1

        for i in range(len(bins)):
            remaining = CAPACITY - (bins[i] + item)

            if remaining >= 0 and remaining < smallest_space:
                smallest_space = remaining
                best_index = i

        if best_index == -1:
            bins.append(item)
        else:
            bins[best_index] += item

    return bins


# -------------------------------------------------
# First Fit Decreasing
# -------------------------------------------------
def first_fit_decreasing(items):
    sorted_items = sorted(items, reverse=True)
    return first_fit(sorted_items)


# -------------------------------------------------
# Display results
# -------------------------------------------------
def show_result(name, bins, lower_bound):
    print(f"\n{name}")
    print("-" * 35)

    for i, load in enumerate(bins, 1):
        print(f"Bin {i}: {load:.1f} / {CAPACITY:.1f}")

    print(f"Bins used       : {len(bins)}")
    print(f"Lower bound     : {lower_bound}")
    print(f"Extra bins      : {len(bins) - lower_bound}")


# Calculate theoretical lower bound
total_weight = sum(weights)
lower_bound = math.ceil(total_weight / CAPACITY)

print("========== BIN PACKING ==========")
print("Items:", weights)
print(f"Total weight: {total_weight:.1f}")
print(f"Bin capacity: {CAPACITY:.1f}")
print(f"Theoretical lower bound: {lower_bound}")

# Apply algorithms
ff = first_fit(weights)
bf = best_fit(weights)
ffd = first_fit_decreasing(weights)

show_result("1. FIRST FIT", ff, lower_bound)
show_result("2. BEST FIT", bf, lower_bound)
show_result("3. FIRST FIT DECREASING", ffd, lower_bound)
```
